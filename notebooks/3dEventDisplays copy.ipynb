```python
def pdg_to_particle(pdg):
    if abs(pdg) == 11:
        return "electron"
    if abs(pdg) == 13:
        return "muon"
    if abs(pdg) == 15:
        return "tau"
    return "other"

long_df["particle"] = long_df["pdg"].apply(pdg_to_particle)
```

```python
def pdg_to_name(pdg):
    try:
        return particle.Particle.from_pdgid(pdg).name
    except Exception:
        return "unknown"

long_df["name"] = long_df["pdg"].apply(pdg_to_name)
```

```python
def isPrim_to_particle(isPrim):
    if isPrim > 0:
        return "Primary"
    else:
        return "Secondary"

long_df["primary"] = long_df["isPrimary"].apply(isPrim_to_particle)
```

```python
def plot_event(index, hit_df, scint_df, truth_df, geom, threshold=0.1, vmin=1e-2, vmax=1e3, hitcolor='black', scintcolor='jet', variable='edep'):

    scint_row = scint_df[scint_df["event_id"] == index].iloc[0]
    hit_row = hit_df[hit_df["event_id"] == index].iloc[0]
    truth_row = truth_df[truth_df["evtID"] == index].iloc[0]

    scint_layerID = scint_row["panelID"]
    scint_colID = scint_row["colID"]
    scint_rowID = scint_row["rowID"]
    scint_edep = scint_row[variable]
    scint_isHoriz = scint_row["isHorizontal"]

    hit_xpos = geom["pixel_Xpos"][0][hit_row["colID"]]
    hit_ypos = geom["pixel_Ypos"][0][hit_row["rowID"]]
    hit_zpos = geom["pixel_Zpos"][0][hit_row["layerID"]]

    scint_xpos = geom["scint_bar_Xcenter"][0][scint_colID]
    scint_ypos = geom["scint_bar_Ycenter"][0][scint_rowID]
    scint_zpos = geom["scint_Zpos"][0][scint_layerID] 

    scint_zedges = []
    for i in range(len(geom["scint_Zpos"][0])):
        scint_zedges.append(geom["scint_Zpos"][0][i] - 2.5)
        scint_zedges.append(geom["scint_Zpos"][0][i] + 2.5)

    # Bar bins are made exactly contiguous (no gaps, no overlaps): the half-width used for each bar's
    # bin is derived from the true center-to-center bar pitch (computed from the actual geometry),
    # instead of a hardcoded 2.5mm half-width. The true pitch is a uniform 10.5mm, not 5mm, so the old
    # hardcoded value left a real 5.5mm dead gap between every bar bin.
    x_centers_sorted = np.sort(geom["scint_bar_Xcenter"][0])
    x_pitch = np.diff(x_centers_sorted).mean()
    scint_xedges = []
    for i in range(len(geom["scint_bar_Xcenter"][0])):
        scint_xedges.append(geom["scint_bar_Xcenter"][0][i] - x_pitch / 2)
        scint_xedges.append(geom["scint_bar_Xcenter"][0][i] + x_pitch / 2)

    y_centers_sorted = np.sort(geom["scint_bar_Ycenter"][0])
    y_pitch = np.diff(y_centers_sorted).mean()
    scint_yedges = []
    for i in range(len(geom["scint_bar_Ycenter"][0])):
        scint_yedges.append(geom["scint_bar_Ycenter"][0][i] - y_pitch / 2)
        scint_yedges.append(geom["scint_bar_Ycenter"][0][i] + y_pitch / 2)


    scint_zedges = sorted(list(set(scint_zedges)))
    scint_xedges = sorted(list(set(scint_xedges)))
    scint_yedges = sorted(list(set(scint_yedges)))

    #     else:
    #         scint_zedges.append((geom["scint_Zpos"][0][i-1] + geom["scint_Zpos"][0][i]) / 2)
    # scint_zedges.append(geom["scint_Zpos"][0][-1] + 5)

    edep_hist_zx = bh.Histogram(
        # bh.axis.Regular(len(geom["scint_Zpos"][0])-1, min(geom["scint_Zpos"][0])-5, max(geom["scint_Zpos"][0])+5),
        bh.axis.Variable(scint_zedges),
        bh.axis.Variable(scint_xedges),
        storage=bh.storage.Weight()
    )

    edep_hist_zy = bh.Histogram(
        bh.axis.Variable(scint_zedges),
        bh.axis.Variable(scint_yedges),
        storage=bh.storage.Weight()
    )


    # print(f"Histogram axis Z: {edep_hist_zx.axes[0].edges}, X: {edep_hist_zx.axes[1].edges}")
    # print("\n")
    # print(f"Histogram axis Z: {edep_hist_zy.axes[0].edges}, Y: {edep_hist_zy.axes[1].edges}")
    # print("\n")

    for i in range(len(scint_xpos)):
        # print(f"Scintillator hit {i}: Layer {scint_layerID[i]}, Col {scint_colID[i]}, Row {scint_rowID[i]}, Edep {scint_edep[i]:.2f} MeV, Horiz {scint_isHoriz[i]}, Position ({scint_xpos[i]:.2f}, {scint_ypos[i]:.2f}, {scint_zpos[i]:.2f})")
        if not scint_isHoriz[i] and scint_layerID[i] >= 0:
            # print(f"Filling edep_hist_zx with Z: {scint_zpos[i]:.2f}, X: {scint_xpos[i]:.2f}, Weight: {scint_edep[i]:.2f}, layerID: {scint_layerID[i]}, colID: {scint_colID[i]}, rowID: {scint_rowID[i]}")
            edep_hist_zx.fill(scint_zpos[i], scint_xpos[i], weight=scint_edep[i])

        if scint_isHoriz[i]:
            # print(f"Filling edep_hist_zy with Z: {scint_zpos[i]:.2f}, Y: {scint_ypos[i]:.2f}, Weight: {scint_edep[i]:.2f}, layerID: {scint_layerID[i]}, colID: {scint_colID[i]}, rowID: {scint_rowID[i]}")
            edep_hist_zy.fill(scint_zpos[i], scint_ypos[i], weight=scint_edep[i])
    # The Z axis is genuinely irregular (5mm real panel bins separated by real physical gaps of varying
    # size -- tungsten plates, air gaps, aluminum walls, Pinpoint-block spacing -- unlike the uniform bar
    # pitch above). A Variable axis can't represent "bin, then a gap with no bin, then another bin": any
    # gap between two real panel edges silently becomes its own bin, which is always empty (fill() only
    # ever lands on an exact panel centre) but renders as a dark zero-content cell under LogNorm, making
    # the real data look sparse/"missing" next to it. Mask those filler bins to NaN so only real
    # scintillator-panel bins are drawn.
    #
    # The real per-panel bin width is always exactly 5mm -- twice the 2.5mm half-width already used
    # above to build scint_zedges -- so it is used directly here rather than inferred from the data
    # (e.g. via the minimum or most-common width): panel spacing is irregular enough that two adjacent
    # panels 9mm apart leave their own narrow ~4mm phantom gap, which is *narrower* than several of the
    # real 5mm panel bins next to it, so neither "smallest width" nor "most common width" reliably picks
    # out only the real bins.
    def mask_phantom_bins_along_z(hist, real_width=5.0, tol=0.01):
        widths = np.diff(hist.axes[0].edges)
        is_phantom = ~np.isclose(widths, real_width, atol=tol)
        hist.view()["value"][is_phantom, :] = np.nan
        return hist

    mask_phantom_bins_along_z(edep_hist_zx)
    mask_phantom_bins_along_z(edep_hist_zy)

    # Make bins with < threshold energy deposition invisible (NaN) so they don't show up in the plot
    edep_hist_zx.view()["value"][edep_hist_zx.view()["value"] < threshold] = np.nan
    edep_hist_zy.view()["value"][edep_hist_zy.view()["value"] < threshold] = np.nan


    fig, ax = plt.subplots(ncols=2, figsize=(32, 8))

    ax[0].plot(hit_zpos, hit_xpos, 'ko', markersize=5, label='Pixel Hits')
    ax[1].plot(hit_zpos, hit_ypos, 'ko', markersize=5, label='Pixel Hits')
    h1 = mplhep.hist2dplot(edep_hist_zx, ax=ax[0], cmap='jet', flow="show", norm=LogNorm(vmin=vmin, vmax=vmax))
    h2 = mplhep.hist2dplot(edep_hist_zy, ax=ax[1], cmap='jet', flow="show", norm=LogNorm(vmin=vmin, vmax=vmax))

    # print(len(set(hit_zpos)))
    # print(len(scint_zedges))

    h1.cbar.set_label(r"$E_{\text{dep}}$ [MeV]")
    h2.cbar.set_label(r"$E_{\text{dep}}$ [MeV]")

    vertex = truth_row[["initX", "initY", "initZ"]].values
    print(f"Truth Vertex: X={vertex[0]:.2f} mm, Y={vertex[1]:.2f} mm, Z={vertex[2]:.2f} mm")
    vertex = truth_row[["initX", "initY", "initZ"]].values
    isCC = True if "CC" in truth_row["processName"] else False
    init_pdg = truth_row["initPDG"]
    init_particle = particle.Particle.from_pdgid(init_pdg).latex_name
    init_energy = truth_row["initE"] / 1000


    fontsize = 24
    title = rf"{'CC' if isCC else 'NC'} ${init_particle}$    $E_\nu$ = {init_energy:.1f} GeV"
    ax[0].set_title(title, loc='left', fontsize=fontsize)
    ax[1].set_title(title, loc='left', fontsize=fontsize)

    ax[0].plot(vertex[2], vertex[0], 'bx', markersize=10, label='Truth Vertex', markeredgewidth=2)
    ax[1].plot(vertex[2], vertex[1], 'bx', markersize=10, label='Truth Vertex', markeredgewidth=2)
    ax[0].legend(loc='upper left', fontsize=fontsize)
    ax[1].legend(loc='upper left', fontsize=fontsize)
    ax[0].set_xlabel("Z (mm)")
    ax[0].set_ylabel("X (mm)")
    ax[1].set_xlabel("Z (mm)")
    ax[1].set_ylabel("Y (mm)")
    ax[0].set_xlim(np.amin(geom["scint_Zpos"][0])-100, np.amax(geom["scint_Zpos"][0]) + 100)
    ax[1].set_xlim(np.amin(geom["scint_Zpos"][0])-100, np.amax(geom["scint_Zpos"][0]) + 100)
    ax[0].set_xlim(-3200, -1650)
    ax[1].set_xlim(-3200, -1650)
    # ax[0].set_ylim(-225 -77, 225-77)
    # ax[1].set_ylim(-225 -77, 225-77)
    fig.tight_layout()

    # for z in geom["scint_Zpos"][0]:
    #     ax[0].axvline(z, color='black', linestyle='--', linewidth=1)
    #     ax[1].axvline(z, color='black', linestyle='--', linewidth=1)
        
```
